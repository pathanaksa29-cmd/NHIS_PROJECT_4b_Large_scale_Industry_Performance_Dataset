```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style='whitegrid')
```

```python
df = pd.read_csv('retail_large_dataset.csv')
df.head()
```

```python
df.info()
df.describe()
```

## 🧹 Data Cleaning

```python
# Convert date
df['order_date'] = pd.to_datetime(df['order_date'])

# Create new columns
df['year'] = df['order_date'].dt.year
df['day_name'] = df['order_date'].dt.day_name()

# Final price
df['final_price'] = df['product_price'] * df['quantity'] * (1 - df['discount_percentage']/100)

# Age group
df['age_group'] = pd.cut(df['age'],
                        bins=[18,25,35,50,60,100],
                        labels=['18-25','26-35','36-50','51-60','60+'])

df.head()
```

## 📈 Univariate Analysis

```python
df['final_price'].describe()
```

```python
# Skewness & Kurtosis
print("Skewness:", df['final_price'].skew())
print("Kurtosis:", df['final_price'].kurt())
```

```python
sns.histplot(df['final_price'], bins=50)
plt.title("Final Price Distribution")
plt.show()
```

👉 Insight: Data is right skewed → few high-value transactions dominate revenue

## 📊 Categorical Analysis

```python
df['product_category'].value_counts().plot(kind='bar')
plt.title("Product Category Distribution")
plt.show()
```

👉 Insight: Electronics category dominates sales

## 🚨 Outlier Detection

```python
sns.boxplot(x=df['final_price'])
plt.title("Outliers in Final Price")
plt.show()
```

👉 Insight: High-value purchases exist but should not be removed

## 🔗 Bivariate Analysis

```python
df[['product_price','final_price','quantity']].corr()
```

```python
sns.heatmap(df[['product_price','final_price','quantity']].corr(), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

👉 Insight: Strong correlation between product_price and final_price

```python
df.groupby('customer_segment')['final_price'].mean().plot(kind='bar')
plt.title("Customer Segment vs Spending")
plt.show()
```

👉 Insight: Premium customers spend more

## 🔥 Multivariate Analysis

```python
df.groupby(['product_category','shipping_type'])['final_price'].mean()
```

👉 Insight: Electronics + fast shipping drives higher revenue

## ✅ Conclusion

- Revenue is right-skewed
- Premium customers drive profits
- Moderate discounts increase sales
- Delivery delays increase returns
